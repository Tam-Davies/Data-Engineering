### Adding Columns

```python
from pyspark.sql import functions as F
df = df.withColumn('tip_percentage', (df['tip_amount'] / df['fare_amount']) *100)
```

```python
df.show()
```

### Dropping a column

```python
df = df.drop('tip_percentage')
```

```python
df.show()
```

### Renaming a column

```python
df.withColumnRenamed('trip_distance', 'distance_per_trip').show()
```

### Using filters 

```python
df.filter(df['fare_amount'] > 50).count()
```

### You can also use filter with select, to show selected columns for the filter rows

```python
df.where(df['fare_amount'] > 50).count()
```

### Selecting Distinct Values from a column like VendorID

```python
df.select('VendorID').distinct().count()
```

### Using orderby to get top fare amount

```python
df.orderBy(df['fare_amount'].desc()).show()
```

### Using Groupby

```python
df.groupBy('VendorId').count().show()
```

### Pyspark Handling Missing Values

```python
py_spark = SparkSession.builder.appName("Second Project").getOrCreate()
py_spark
```

```python
df_pyspark = py_spark.read.option('header', 'true').option('inferSchema', 'true').csv('2015.csv')
df_pyspark.show()
```

### Data Summary with GroupBy and Agg function

```python
data_summary = df_pyspark.groupBy('VendorId').agg(
    F.count('*').alias('total_trips'),
    F.avg('fare_amount').alias('average_fare'),
    F.sum('tip_amount').alias('total_tips'))

data_summary.show()
```

### Counting Null values

```python
df_pyspark.na.drop().count()
```

when dropping values the drop function takes argument how with parameters like any, thresh and numeric parameter for at least non null values, subset and na records in a given column

### Filling Missing values

```python
df_pyspark.na.fill('Missing_Values').show()
# df_pyspark.na.fill('Missing_Values', ['columns']).show()
```

### Using SQL in Spark

```python
df_pyspark.createOrReplaceTempView("taxi_trips")
```

```python
py_spark.sql("SELECT * FROM taxi_trips LIMIT 10").show()
```
