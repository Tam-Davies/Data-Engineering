```python
from bs4 import BeautifulSoup
import requests
headers = {
    "User-Agent": "DataEngineeringPracticeBot/1.0 (your.email@example.com; personal learning project)"
}
def extract_data():
    url = 'https://en.wikipedia.org/wiki/List_of_largest_companies_in_the_United_States_by_revenue'

    page = requests.get(url, headers=headers)

    soup = BeautifulSoup(page.text, 'html.parser')
    return soup

if __name__ == '__main__':
    page = extract_data()
    print(page)
```

```python
# Getting Table Titles
table_title = page.find_all('th')
table_title
```

```python
table_titles = [title.text.strip() for title in table_title]
print(table_titles)
```

### FOR JUST ONE TABLE

```python
table = page.find_all('table')[0]
```

```python
table_title = table.find_all('th')

table_titles = [title.text.strip() for title in table_title]
print(table_titles)
```

```python
column_data = table.find_all('tr')
print(column_data)
```

```python
import pandas as pd


df = pd.DataFrame(columns=table_titles)
for rows in column_data[1:]:
    row_data = rows.find_all('td')
    individual_data = [data.text.strip() for data in row_data]
    length = len(df)
    df.loc[length] = individual_data

df.head(10)
```
